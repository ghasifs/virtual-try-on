```python
from ultralytics import YOLO
```

```python

# Load the YOLO11n-pose model
model = YOLO("yolo11n-pose.pt")  # Pretrained model

# Fine-tune on the COCO dataset
model.train(
    data="coco_keypoints.yaml",  # Dataset configuration
    epochs=5,                  # Number of epochs
    imgsz=640,                  # Image size
    batch=16,                   # Batch size
    device=0,                   # GPU device
    workers=4,                  # Number of data loading workers
    pretrained=True,            # Use pretrained weights
    save_period=5,              # Save model every 5 epochs
)
```

```python
# Load the fine-tuned model
model = YOLO("runs/pose/train2/weights/best.pt")

# Perform inference on a test image
results = model.predict(source="test.jpg", imgsz=640, save=True)
#results = model.predict(source="datasets/images/test", imgsz=640, save=True)

# Display the results
#results[0].show()  # Opens an image with keypoints drawn


#model.val()  # Evaluate the model on the COCO validation set
```
